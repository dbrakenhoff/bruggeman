# 244.02: Circular basin with radial flow

Circular basin with a water level that differs from the surrounding polder.
Different resistances for basin and polder are represented with leakage resistances `c1` and `c2`.

```python
import matplotlib.pyplot as plt
import numpy as np

from bruggeman.flow2d import bruggeman_244_02
```

View the function.

```python
bruggeman_244_02
```

View the docstring to get a description of the input parameters.

```python
help(bruggeman_244_02)
```

Define aquifer and basin parameters.

```python
R = 700.0  # m
h = 3  # m
k = 15.0  # m/d
D = 20.0  # m
c1 = 80.0  # d (inside basin)
c2 = 300.0  # d (outside basin)
```

Compute and plot the radial head profile.

```python
r = np.linspace(0.0, 4 * R, 400)
phi = bruggeman_244_02(r, R, h, k, D, c1, c2)

fig, ax = plt.subplots(figsize=(8, 3), layout="tight")
ax.plot(r, phi, label="Head profile")
ax.axvline(R, color="k", linestyle="--", linewidth=1.0, label="Basin edge (r = R)")
ax.grid(True)
ax.set_xlabel("Radial distance [m]")
ax.set_ylabel("Head [m]")
plt.legend(loc=(0, 1), frameon=False, ncol=2, fontsize="small");
```

```python
flux = np.where(r <= R, (phi - h) / c1, phi / c2)

r_inside = r[r <= R]
flux_inside = flux[r <= R]
area_inside = np.pi * (r_inside**2 - np.concatenate(([0.0], r_inside[:-1])) ** 2)
cumulative_flux_inside = np.cumsum(flux_inside * area_inside)
total_flux_inside = cumulative_flux_inside[-1]

r_outside = r[r > R]
flux_outside = flux[r > R]
area_outside = np.pi * (r_outside**2 - np.concatenate(([R], r_outside[:-1])) ** 2)
cumulative_flux_outside = np.cumsum(flux_outside * area_outside)
total_flux_outside = cumulative_flux_outside[-1]

fig, ax = plt.subplots(figsize=(8, 3), layout="tight")
ax.plot(r, flux, label="Flux profile")
ax.axvline(R, color="k", linestyle="--", linewidth=2.0, label="Basin edge (r = R)")
ax.grid(True)
ax.set_xlabel("Radial distance [m]")
ax.set_ylabel("Flux [m/day]")
ax.legend(loc=(0, 1), frameon=False, ncol=7, fontsize="small")

print(f"Total flux inside the basin: {total_flux_inside:.0f} m$^3$/d")
print(f"Total flux outside the basin: {total_flux_outside:.0f} m$^3$/d")
```
