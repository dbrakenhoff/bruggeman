# 244.02: Circular basin with radial flow

Circular basin with a water level that differs from the surrounding polder.
Different permeable layers for basin and polder are represented with leakage resistances `c1` and `c2`.

```python
import matplotlib.pyplot as plt
import numpy as np

from bruggeman.flow2d import bruggeman_244_02
```

View the function.

```python
bruggeman_244_02
```

View the docstring to get a description of the input parameters.

```python
help(bruggeman_244_02)
```

Define aquifer and basin parameters.

```python
R = 700.0  # m
h = 3  # m
k = 15.0  # m/d
D = 20.0  # m
c1 = 80.0  # d (inside basin)
c2 = 300.0  # d (outside basin)
```

Compute and plot the radial head profile.

```python
r = np.linspace(0.0, 4 * R, 400)
phi = bruggeman_244_02(r, R, h, k, D, c1, c2)

fig, ax = plt.subplots(figsize=(8, 3), layout="tight")
ax.plot(r, phi, label="Head profile")
ax.axvline(R, color="k", linestyle="--", linewidth=1.0, label="Basin edge (r = R)")
ax.grid(True)
ax.set_xlabel("Radial distance r (m)")
ax.set_ylabel("Head (m)")
ax.set_title("Bruggeman 244.02")
ax.legend();
```
