```python
t = 1 # time, d
h = -2  # change in head, m
k = 20.0  # hydraulic conductivity, m/d
D = 50.0  # thickness of aquifer, m
```

```python
x =  linspace(0, 2000, 100)
c = linspace(100, 500, 5) # resitance, d
plt.figure(figsize=(10, 3))
for ci in c:
    ha = bruggeman_123_33(x, h, k, D, ci)
    plt.plot(x, ha, label=f"c={ci:.0f} days")
plt.legend(loc=(0, 1), frameon=False, ncol=7, fontsize="small")
plt.grid()
plt.ylabel("Drawdown [m]")
plt.xlabel("Distance [m]")
plt.show()
```
