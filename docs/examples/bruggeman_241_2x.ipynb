# 241.2x Pumping well at the centre of circular islands 

#### 241.25: Pumping well at the centre of a circular island with open vertical boundary

```python
import matplotlib.pyplot as plt
import numpy as np

from bruggeman.flow2d import bruggeman_241_25,bruggeman_241_27
```

View the function.

```python
bruggeman_241_25
```

View the docstring to get a description of the input parameters.

```python
help(bruggeman_241_25)
```

Define aquifer and basin parameters.

```python
R = 100.0  # m
Q0 = -1000  
k = 100  
D = 10
c = 100
```

Compute and plot the radial head profile.

```python
r = np.linspace(0.0, R, 400)
Q0 = np.linspace(-500,-1000,3)
fig, ax = plt.subplots(figsize=(8, 3), layout="tight")
for Qi in Q0:
    ha = bruggeman_241_25(r, R, Qi, k, D, c)
    plt.plot(r, ha, label=f"c={Qi:.0f} m$^3$/day")
ax.grid(True)
ax.set_xlabel("Radial distance r [m]")
ax.set_ylabel("Head [m]")
plt.legend(loc=(0, 1), frameon=False, ncol=3, fontsize="small")
```

#### 241.27: Steady-state solution for a pumping well at the centre of a circular island with closed vertical boundary at r = R. Constant discharge.

View the function.

View the docstring to get a description of the input parameters.

Define aquifer and basin parameters.

```python
R = 1000.0  # m
Q0 = -1000  
k = 100  
D = 10
c = 100
```

Compute and plot the radial head profile.

```python
r = np.linspace(0.0, R, 400)
Q0 = np.linspace(-500,-1000,3)
fig, ax = plt.subplots(figsize=(8, 3), layout="tight")
for Qi in Q0:
    ha = bruggeman_241_27(r, R, Qi, k, D, c)
    plt.plot(r, ha, label=f"c={Qi:.0f} m$^3$/day")
ax.grid(True)
ax.axvline(R, color="k", linestyle="--", linewidth=1.0, label="Basin edge (r = R)")
ax.set_xlabel("Radial distance r [m]")
ax.set_ylabel("Head [m]")
plt.legend(loc=(0, 1), frameon=False, ncol=4, fontsize="small")
```
