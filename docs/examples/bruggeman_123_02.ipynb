# 123.02: Sudden rise of the surface water level which is kept constant thereafter

```python
import matplotlib.pyplot as plt
import numpy as np

from bruggeman.flow1d import bruggeman_123_02
```

View the function (this will be rendered in LaTeX if `sympy` is installed).

```python
bruggeman_123_02
```

View the docstring to get a description of the input parameters.

```python
help(bruggeman_123_02)
```

Define some aquifer parameters.

```python
h = -1.0  # m, rise in head
k = 5.0  # m/d, hydraulic conductivity
D = 10.0  # m  # thickness aquifer
Ss = 1e-3 / D  # m^-1, specific storage coeffecient
```

```python
x = np.linspace(0, 100, 101)
t = [1,2,5,10,30,60,120,360,720,1440] # minutes

plt.figure(figsize=(10, 3))
plt.figure(figsize=(10, 3))
for ti in t:
    t_days = ti / 60 / 24 # convert to hours
    ha = bruggeman_123_02(x, t_days, h, k, D, Ss)
    plt.plot(x, ha, label=f"t={ti:.0f} min.")
plt.legend(loc=(0, 1), frameon=False, ncol=5, fontsize="small")
plt.xlabel("x [m]")
plt.ylabel("drawdown [m]")
plt.grid()
```
