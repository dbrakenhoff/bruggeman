# 123.02: Sudden rise of the surface water level which is kept constant thereafter

```python
import matplotlib.pyplot as plt
import numpy as np
import timflow.transient as tft

from bruggeman import bruggeman_123_02
```

Define some aquifer parameters.

```python
h = -1.0  # m, rise in head
k = 5.0  # m/d, hydraulic conductivity
D = 10.0  # m  # thickness aquifer
Ss = 1e-3 / D  # m^-1, specific storage coeffecient
```

Compare to the solution computed with timflow.

```python
# timflow solution
ml = tft.ModelXsection(naq=1, tmin=1e-8, tmax=1e2)
aq = tft.XsectionMaq(
    ml, x1=-np.inf, x2=np.inf, kaq=[k], z=[0, -D], Saq=[Ss], topboundary="phreatic"
)
ls = tft.River1D(ml, xls=0.0, tsandh=[(0, h)])
ml.solve(silent=True)
```

```python
x = np.linspace(0, 100, 101)
t = np.array([1, 2, 5, 10, 30, 60, 120, 360, 720, 1440])  # minutes

h_ml = ml.headalongline(x[::10], 0.0, t / 60 / 24)

plt.figure(figsize=(10, 3))
for i, ti in enumerate(t):
    t_days = ti / 60 / 24  # convert to hours
    ha = bruggeman_123_02(x, t_days, h, k, D, Ss)
    (ph,) = plt.plot(x, ha, label=f"t={ti:.0f} min.")
    plt.plot(x[::10], h_ml[0, i, :], marker="x", ls="none", color=ph.get_color())
plt.plot([], [], "kx", label="timflow")
plt.legend(loc=(0, 1), frameon=False, ncol=6, fontsize="small")
plt.xlabel("x [m]")
plt.ylabel("drawdown [m]")
plt.grid();
```
