# bruggeman_215_03: Theiss solution for constant discharge to a confined aquifer.


```python
import matplotlib.pyplot as plt
import numpy as np

from bruggeman.flow2d import bruggeman_215_03
```

View the function.

```python
bruggeman_215_03
```

View the docstring to get a description of the input parameters.

```python
help(bruggeman_215_03)
```

Define aquifer and basin parameters.

```python
R = 100.0  # m
t = 1  # m
Q0 = -1000  
k = 100  
D = 10
S = 0.001
```

Compute and plot the radial head profile.

```python
r = np.linspace(0.0, 10 * R, 400)
Q0 = np.linspace(-500,-1000,3)
fig, ax = plt.subplots(figsize=(8, 3), layout="tight")
for Qi in Q0:
    ha = bruggeman_215_03(r, t, Qi, k, D, S)
    plt.plot(r, ha, label=f"c={Qi:.0f} m$^3$/day")
ax.grid(True)
ax.set_xlabel("Radial distance r [m]")
ax.set_ylabel("Head [m]")
plt.legend(loc=(0, 1), frameon=False, ncol=7, fontsize="small")
```
