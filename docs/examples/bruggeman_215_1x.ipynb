# 215.1x Radial flow to a line source (pumping well or drain) 

#### 215.13: Hantush solution for constant discharge to a leaky confined aquifer.


```python
import matplotlib.pyplot as plt
import numpy as np

from bruggeman.flow2d import bruggeman_215_13, bruggeman_215_14
```

View the function.

```python
bruggeman_215_13
```

View the docstring to get a description of the input parameters.

```python
help(bruggeman_215_13)
```

Define aquifer and basin parameters.

```python
R = 100.0  # m
t = 1  # m
Q0 = -1000  
k = 100  
D = 10
c = 100
eta = 10
```

Compute and plot the radial head profile.

```python
r = np.linspace(0.0, 5 * R, 400)
Q0 = np.linspace(-500,-1000,3)
fig, ax = plt.subplots(figsize=(8, 3), layout="tight")
for Qi in Q0:
    ha = bruggeman_215_13(r, t, Qi, k, D, c, eta)
    plt.plot(r, ha, label=f"c={Qi:.0f} m$^3$/day")
ax.grid(True)
ax.set_xlabel("Radial distance r [m]")
ax.set_ylabel("Head [m]")
ax.set_ylim(-1, 0)
plt.legend(loc=(0, 1), frameon=False, ncol=7, fontsize="small")
```
