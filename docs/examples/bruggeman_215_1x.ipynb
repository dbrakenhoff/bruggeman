# 215.1x Radial flow to a line source (pumping well or drain) 

#### 215.13: Hantush solution for constant discharge to a leaky confined aquifer.


```python
import matplotlib.pyplot as plt
import numpy as np
import timflow.steady as tfs
import timflow.transient as tft

from bruggeman import bruggeman_215_13, bruggeman_215_14
```

View the function.

```python
bruggeman_215_13
```

View the docstring to get a description of the input parameters.

```python
help(bruggeman_215_13)
```

Define aquifer and basin parameters.

```python
R = 100.0  # m
t = 1  # m
Q0 = -1000
k = 100
D = 10
c = 100
eta = 10
```

Compare to the solution computed with timflow.

```python
def timflow_hantush(Q0, r, t):
    ml = tft.ModelMaq(
        kaq=[k], z=[1, 0, -D], c=[c], Saq=[1 / (eta * c) / D], topboundary="semi"
    )
    tft.Well(ml, xw=0.0, yw=0.0, tsandQ=[(0, -Q0)])
    ml.solve(silent=True)
    return ml.headalongline(r, 0.0, t)
```

Compute and plot the radial head profile.

```python
r = np.linspace(0.0, 5 * R, 400)
Q0 = np.linspace(-500, -1000, 3)


fig, ax = plt.subplots(figsize=(8, 3), layout="tight")
for Qi in Q0:
    ha = bruggeman_215_13(r, t, Qi, k, D, c, eta)
    (ph,) = plt.plot(r, ha, label=f"$Q_0$={Qi:.0f} m$^3$/day")
    h_ml = timflow_hantush(Qi, r[::40], t)
    plt.plot(r[::40], h_ml[0, 0], marker="x", ls="none", color=ph.get_color())
plt.plot([], [], "kx", label="timflow")
ax.grid(True)
ax.set_xlabel("Radial distance r [m]")
ax.set_ylabel("Head [m]")
ax.set_ylim(-1, 0)
plt.legend(loc=(0, 1), frameon=False, ncol=7, fontsize="small");
```

#### 215.14: De Glee, steady state abstraction of water from a line source

View the function.

```python
bruggeman_215_14
```

View the docstring to get a description of the input parameters.

```python
help(bruggeman_215_14)
```

Define aquifer and basin parameters.

```python
R = 100.0  # m
Q0 = -1000
k = 100
D = 10
c = 100
```

Compare to the solution computed with timflow.

```python
def timflow_deglee(Q0, r):
    ml = tfs.ModelMaq(kaq=[k], z=[1, 0, -D], c=[c], topboundary="semi", hstar=0.0)
    tfs.Well(ml, xw=0.0, yw=0.0, Qw=-Q0)
    ml.solve(silent=True)
    return ml.headalongline(r, 0.0)
```

Compute and plot the radial head profile.

```python
r = np.linspace(0.0, 5 * R, 400)
Q0 = np.linspace(-500, -1000, 3)
fig, ax = plt.subplots(figsize=(8, 3), layout="tight")
for Qi in Q0:
    ha = bruggeman_215_14(r, Qi, k, D, c)
    (ph,) = plt.plot(r, ha, label=f"c={Qi:.0f} m$^3$/day")
    h_ml = timflow_deglee(Qi, r[1::40])
    plt.plot(r[1::40], h_ml[0], marker="x", ls="none", c=ph.get_color())
plt.plot([], [], "kx", label="timflow")
ax.grid(True)
ax.set_xlabel("Radial distance r [m]")
ax.set_ylabel("Head [m]")
ax.set_ylim(-1, 0)
plt.legend(loc=(0, 1), frameon=False, ncol=4, fontsize="small");
```
